```python
from scipy.spatial.distance import euclidean

# Kasus 1: Tinggi dalam milimeter (mm)
X_train = np.array([[1700, 0], [1600, 1]])
X_new = np.array([[1640, 0]])

jarak_mm = [euclidean(X_new[0], d) for d in X_train]
print(f'Jarak dalam mm: {jarak_mm}')
```

```python
# Kasus 2: Tinggi dalam meter (m)
X_train = np.array([[1.7, 0], [1.6, 1]])
X_new = np.array([[1.64, 0]])

jarak_m = [euclidean(X_new[0], d) for d in X_train]
print(f'Jarak dalam m: {jarak_m}')
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
```

```python
# Tinggi dalam milimeter
X_train = np.array([[1700, 0], [1600, 1]])
X_train_scaled = ss.fit_transform(X_train)
print(f'X_train_scaled:\n{X_train_scaled}\n')

X_new = np.array([[1640, 0]])
X_new_scaled = ss.transform(X_new)
print(f'X_new_scaled: {X_new_scaled}\n')

jarak = [euclidean(X_new_scaled[0], d) for d in X_train_scaled]
print(f'jarak: {jarak}')
```
