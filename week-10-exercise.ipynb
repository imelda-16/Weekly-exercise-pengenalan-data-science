```python
import pandas as pd
import os

possible_paths = [
    r'C:\Users\Imelda\weekly-exercise-pengenalan-data-science\weekly-exercise-pengenalan-data-science\dataset\SMSSpamCollection',
    r'C:\Users\Imelda\weekly-exercise-pengenalan-data-science\weekly-exercise-pengenalan-data-science\dataset\SMSSpamCollection.txt',
    r'C:\Users\Imelda\weekly-exercise-pengenalan-data-science\weekly-exercise-pengenalan-data-science\dataset\SMSSpamCollection.csv',
    # Alternative: Use relative path if file is in same directory as notebook
    'SMSSpamCollection',
    'SMSSpamCollection.txt',
    'dataset/SMSSpamCollection',
    'dataset/SMSSpamCollection.txt'
]


for path in possible_paths:
    if os.path.exists(path):
        print(f"Found file at: {path}")
        df = pd.read_csv(path, 
                         sep='\t', 
                         header=None, 
                         names=['label', 'sms'])
        print(df.head())
        break
else:
    print("File not found. Please check:")
    print("1. The file path is correct")
    print("2. The file exists at that location") 
    print("3. You have permission to access the file")
    print("4. Try using forward slashes: C:/Users/Imelda/...")
```

```python
df['label'].value_counts()
```

```python
from sklearn.preprocessing import LabelBinarizer

X = df['sms'].values
y = df['label'].values

lb = LabelBinarizer()
y = lb.fit_transform(y).ravel()
lb.classes_
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y,
                                                    test_size=0.25,
                                                    random_state=0)

print(X_train, '\n')
print(y_train)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words='english')

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print(X_train_tfidf)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train_tfidf, y_train)
y_pred = model.predict(X_test_tfidf)

for pred, sms in zip(y_pred[:5], X_test[:5]):
    print(f'PRED: {pred} - SMS: {sms}\n')
```

```python
from sklearn.metrics import confusion_matrix

matrix = confusion_matrix(y_test, y_pred)
matrix
```

```python
tn, fp, fn, tp = matrix.ravel()

print(f'TN: {tn}')
print(f'FP: {fp}')
print(f'FN: {fn}')
print(f'TP: {tp}')
```

```python
import matplotlib.pyplot as plt

plt.matshow(matrix)
plt.colorbar()

plt.title('Confusion Matrix')
plt.ylabel('True label')
plt.xlabel('Predicted label')
plt.show()
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_test)
```

```python
from sklearn.metrics import roc_curve, auc

prob_estimates = model.predict_proba(X_test_tfidf)

fpr, tpr, threshhold = roc_curve(y_test, prob_estimates[:, 1])
nilai_auc = auc(fpr, tpr)

plt.plot(fpr, tpr, 'b', label=f'AUC={nilai_auc}')
plt.plot([0, 1], [0, 1], 'r--', label='Random Classifier')

plt.title('ROC: Receiver Operating Characteristic')
plt.xlabel('Fallout or False Positive Rate')
plt.ylabel('Recall or True Positive Rate')
plt.legend()
plt.show()
```
