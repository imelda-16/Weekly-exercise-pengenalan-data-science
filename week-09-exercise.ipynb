```python
corpus = [
    'Linux has been around since the mid-1990s.',
    'Linux distributions include the Linux kernel.',
    'Linux is one of the most prominent open-source software.'
]

corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
vectorized_X = vectorizer.fit_transform(corpus).todense()
vectorized_X
```

```python
vectorizer.get_feature_names_out()
```

```python
import numpy as np
from sklearn.metrics.pairwise import euclidean_distances

for i in range(len(vectorized_X)):
    for j in range(i, len(vectorized_X)):
        if i == j:
            continue
        
        vec_i = np.asarray(vectorized_X[i])
        vec_j = np.asarray(vectorized_X[j])
        
        jarak = euclidean_distances(vec_i, vec_j)
        print(f'Jarak dokumen {i+1} dan {j+1}: {jarak}')
```

```python
corpus = [
    'the house had a tiny little mouse',
    'the cat saw the mouse',
    'the mouse ran away from the house',
    'the cat finally ate the mouse',
    'the end of the mouse story'
]

corpus
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words='english')
response = vectorizer.fit_transform(corpus)
print(response)
```

```python
vectorizer.get_feature_names_out()
```

```python
response.todense()
```

```python
import pandas as pd

df = pd.DataFrame(response.todense().T,
                 index=vectorizer.get_feature_names_out(),  # Changed from get_feature_names() to get_feature_names_out()
                 columns=[f'D{i+1}' for i in range(len(corpus))])
df
```
